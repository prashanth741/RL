```python
def env_state():
    states = ["S0", "S1"]
    return states

states = env_state()
print("States:", states)

def env_action():
    actions = ["up", "down", "left", "right"]
    return actions

actions = env_action()
print("Actions:", actions)

def env_df():
    gamma = 0.9
    return gamma

gamma = env_df()
print("Discount Factor:", gamma)

def env_rewards():
    rewards = {
        "Move": -1,
        "Wall": -1,
        "Goal": 10
    }
    return rewards

rewards = env_rewards()
print("Rewards:", rewards)

def env_transition():
    transition = {
        "S0": {
            "up": "S0",
            "down": "S1",
            "left": "S0",
            "right": "S0"
        },

        "S1": {
            "up": "S0",
            "down": "S1",
            "left": "S1",
            "right": "Goal"
        }
    }
    return transition

transitions = env_transition()
print("Transitions:", transitions)

def rewards_cal(next_state):

    if next_state == "Goal":
        reward = rewards["Goal"]

    elif next_state == "Wall":
        reward = rewards["Wall"]

    else:
        reward = rewards["Move"]

    return reward
state = "S1"
action = "up"

next_state = transitions[state][action]

print("Next State:", next_state)

reward = rewards_cal(next_state)

print("Reward:", reward)

def env_policy():
    policy = {
        "S0": "up",
        "S1": "up"
    }
    return policy

policy = env_policy()

print("\nInitial Policy:", policy)
V = {
    "S0": 0.0,
    "S1": 0.0,
    "Goal": 0.0
}

theta = 1e-8
max_iterations = 10000

def policy_evaluation(policy, V):

    for iteration in range(max_iterations):

        old_V = V.copy()
        new_V = old_V.copy()

        for state in states:


            action = policy[state]

            next_state = transitions[state][action]
            r = rewards_cal(next_state)

            new_V[state] = r + gamma * old_V[next_state]


        delta = max(
            abs(new_V[s] - old_V[s])
            for s in states
        )

        V = new_V

        if delta < theta:
            break

    return V

def calculate_q(state, action, V):

    next_state = transitions[state][action]

    r = rewards_cal(next_state)

    return r + gamma * V[next_state]



while True:

    V = policy_evaluation(policy, V)


    improved_policy = {}

    for state in states:

        q_values = {
            action: calculate_q(state, action, V)
            for action in actions
        }

        improved_policy[state] = max(
            q_values,
            key=q_values.get
        )

    if improved_policy == policy:
        break

    policy = improved_policy



print("\nOptimal State Values:")

for state, value in V.items():
    print(state, ":", round(value, 4))

print("\nOptimal Policy:")

for state, action in policy.items():
    print(state, ":", action)
```
