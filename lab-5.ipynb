```python
import gymnasium as gym
import numpy as np

def run_random_agent(env_name, num_episodes=100):

    print(f"\n--- Running Random Agent on {env_name} ---")

    try:
        env = gym.make(env_name)
    except Exception as e:
        print(f"Error creating environment {env_name}: {e}")
        return

    all_episode_rewards = []

    for episode in range(num_episodes):
        observation, info = env.reset()
        episode_reward = 0
        terminated = False
        truncated = False

        while not terminated and not truncated:
            action = env.action_space.sample()
            observation, reward, terminated, truncated, info = env.step(action)
            episode_reward += reward

        all_episode_rewards.append(episode_reward)

    env.close()

    total_reward = np.sum(all_episode_rewards)
    average_reward = np.mean(all_episode_rewards)

    print(f"Total Reward over {num_episodes} episodes: {total_reward:.2f}")
    print(f"Average Reward per episode over {num_episodes} episodes: {average_reward:.2f}")



run_random_agent('CartPole-v1')


run_random_agent('MountainCar-v0')


run_random_agent('FrozenLake-v1', num_episodes=1000)
```
